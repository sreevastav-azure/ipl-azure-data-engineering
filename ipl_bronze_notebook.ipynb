```python
# =============================
# IPL BRONZE NOTEBOOK
# =============================

# COMMAND ----------
# Read matches data
df_matches = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("abfss://ipl-bronze@adlsstoragertp9.dfs.core.windows.net/matches/matches.csv")

df_matches.display()
df_matches.printSchema()

# COMMAND ----------
# Write to bronze delta table
df_matches.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("ipl_bronze.cricket.matches")

print("matches done ✅")

# COMMAND ----------
# Read deliveries data
df_deliveries = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("abfss://ipl-bronze@adlsstoragertp9.dfs.core.windows.net/deliveries/deliveries.csv")

df_deliveries.display()
df_deliveries.printSchema()

# COMMAND ----------
# Write to bronze delta table
df_deliveries.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("ipl_bronze.cricket.deliveries")

print("deliveries done ✅")
```
